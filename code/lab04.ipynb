#### Проверка:

```python
import os
import sys
import warnings
import logging
import torchaudio

warnings.filterwarnings("ignore", category=FutureWarning, module="diffusers")
logging.getLogger('cosyvoice').setLevel(logging.WARNING)

if cosyvoice_path not in sys.path:
    sys.path.append(cosyvoice_path)
if matcha_path not in sys.path:
    sys.path.append(matcha_path)

from cosyvoice.cli.cosyvoice import CosyVoice

model_dir = os.path.join(cosyvoice_path, 'pretrained_models', 'CosyVoice-300M')

if os.path.exists(model_dir):
    cosyvoice = CosyVoice(model_dir)
    
    text_to_speak = "Проверка синтеза речи."
    output_dir = "output"
    os.makedirs(output_dir, exist_ok=True)

    try:
        for i, j in enumerate(cosyvoice.inference_sft(text_to_speak, 'English female')):
            output_path = os.path.join(output_dir, 'result.wav')
            torchaudio.save(output_path, j['tts_speech'], 22050)
            print(f"Success: {output_path}")
    except Exception as e:
        print(f"Error: {e}")
else:
    print(f"Not found: {model_dir}")
```
